```python
import torch
from cuml.cluster import KMeans

x = torch.randn(20, 5).to('cuda:7')

kmeans = KMeans(n_clusters=3)

cluster = kmeans.fit_predict(x) 

cluster
```

```python
from torchmetrics.clustering import AdjustedRandScore, NormalizedMutualInfoScore
from torchmetrics.functional.clustering import adjusted_rand_score, normalized_mutual_info_score

preds = torch.tensor([0, 0, 1, 1, 2, 2])
target = torch.tensor([0, 0, 1, 2, 2, 2])

# 2. 初始化指标对象
# ARI 范围 [-1, 1], 1 为完美匹配
# ari_metric = AdjustedRandScore()

# NMI 范围 [0, 1], 1 为完美匹配
# nmi_metric = NormalizedMutualInfoScore()

# 3. 计算结果
ari_score = adjusted_rand_score(preds, target)
nmi_score = normalized_mutual_info_score(preds, target)

print(f"ARI: {ari_score.item():.4f}")
print(f"NMI: {nmi_score.item():.4f}")
```
